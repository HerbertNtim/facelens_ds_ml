```python
cap = cv2.VideoCapture('')

while True:
    ret, frame = cap.read()

    if ret == False:
        break
    pred_img, pred_dict = genderRecognitionPipeline(frame,path=False)

    cv2.imshow('prediction', pred_img)
    if cv2.waitKey(1) == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
